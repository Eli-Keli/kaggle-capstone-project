### 🎯 **The Problem**

**Kenya has over 2.2 million people with disabilities** (KNBS, 2019). In Nairobi, they face significant barriers:

- ❌ **No centralized database** of accessible facilities
- ❌ **Unclear accessibility information** at most clinics and offices
- ❌ **Transportation challenges** without accessible routes
- ❌ **Cost barriers** with limited free/subsidized options

**Our Goal:** Create a comprehensive, structured dataset that AI agents can use to help people with disabilities find accessible services.

## 📐 Dataset Schema Design

Our dataset is structured around **Facility** entities with comprehensive accessibility information.

```python
# Load the datasets
clinics_df = pd.read_csv('data/nairobi_clinics_accessible.csv')
social_df = pd.read_csv('data/nairobi_social_services_accessible.csv')

print("📋 DATASET SCHEMA")
print("="*60)
print("\n🏥 Clinics Dataset:")
print(f"   Columns: {list(clinics_df.columns)}")
print(f"   Records: {len(clinics_df)}")

print("\n🏢 Social Services Dataset:")
print(f"   Columns: {list(social_df.columns)}")
print(f"   Records: {len(social_df)}")

print("\n✨ Key Fields:")
print("   • name, category, subcounty, ward")
print("   • disability_support (target disabilities)")
print("   • accessibility_features (detailed list)")
print("   • accessibility_score (1-10 rating)")
print("   • cost_estimate, hours_of_operation")
print("   • contact_phone, latitude, longitude")
```

## 🏥 Sample Facilities

Let's examine some real examples from our dataset.

```python
# Display detailed information for a sample facility
sample = clinics_df.iloc[0]

print("🏥 SAMPLE FACILITY: " + sample['name'])
print("="*60)
for col in clinics_df.columns:
    value = sample[col]
    print(f"{col.upper()}: {value}")
print("="*60)

print(f"\n💡 This facility scores {sample['accessibility_score']}/10 for accessibility")
print(f"   Supports: {sample['disability_support']}")
print(f"   Key features: {sample['accessibility_features'][:100]}...")
```

## ♿ Accessibility Scoring Methodology

Each facility receives a score from 1-10 based on multiple factors.

```python
# Combine datasets and analyze accessibility scores
all_facilities = pd.concat([clinics_df, social_df], ignore_index=True)

print("♿ ACCESSIBILITY SCORING ANALYSIS")
print("="*60)
print(f"\n📊 Score Distribution:")
print(all_facilities['accessibility_score'].value_counts().sort_index().to_string())

print(f"\n📈 Statistics:")
print(f"   Mean Score: {all_facilities['accessibility_score'].mean():.2f}/10")
print(f"   Median Score: {all_facilities['accessibility_score'].median():.2f}/10")
print(f"   Highly Accessible (8-10): {len(all_facilities[all_facilities['accessibility_score'] >= 8])} facilities")
print(f"   Moderately Accessible (5-7): {len(all_facilities[(all_facilities['accessibility_score'] >= 5) & (all_facilities['accessibility_score'] < 8)])} facilities")
print(f"   Low Accessibility (1-4): {len(all_facilities[all_facilities['accessibility_score'] < 5])} facilities")

print(f"\n💡 Scoring Criteria:")
print("   • Physical accessibility (ramps, elevators, wide doorways)")
print("   • Disability-specific support (sign language, braille, audio)")
print("   • Staff training on disability awareness")
print("   • Accessible restrooms and parking")
print("   • Service affordability and availability")
```

## 🗺️ Geographic Coverage

Analyzing the distribution of facilities across Nairobi's subcounties.

```python
# Geographic analysis
coverage = all_facilities.groupby('subcounty').agg({
    'name': 'count',
    'accessibility_score': 'mean',
    'category': lambda x: x.value_counts().index[0]
}).round(2)
coverage.columns = ['Facility Count', 'Avg Accessibility', 'Primary Category']
coverage = coverage.sort_values('Facility Count', ascending=False)

print("🗺️ GEOGRAPHIC COVERAGE BY SUBCOUNTY")
print("="*60)
print(coverage.to_string())
print("="*60)

print(f"\n📍 Coverage Summary:")
print(f"   Total Subcounties: {all_facilities['subcounty'].nunique()}")
print(f"   Total Wards: {all_facilities['ward'].nunique()}")
print(f"   Most Covered: {coverage.index[0]} ({coverage.iloc[0]['Facility Count']} facilities)")
print(f"   Least Covered: {coverage.index[-1]} ({coverage.iloc[-1]['Facility Count']} facilities)")

print(f"\n⚠️ Gaps Identified:")
print("   • Eastern Nairobi (Embakasi) could use more facilities")
print("   • Need more social service offices in informal settlements")
print("   • Opportunity for mobile clinics in underserved areas")
```
